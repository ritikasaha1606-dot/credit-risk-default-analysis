```python
import pandas as pd
df = pd.read_csv('credit_risk_dataset.csv')
df.head()
```

```python
df.info()
df.describe()
```

```python
import sqlite3
conn = sqlite3.connect('credit_risk.db')
# now we need to form an actual SQL database
df.to_sql('credit_risk', conn, if_exists='replace', index=False)
```

```python
query = """
SELECT COUNT(*)
FROM credit_risk
WHERE person_age > 100
"""
query_results = pd.read_sql_query(query, conn)
query_results
```

```python
query = """
SELECT *
FROM credit_risk
WHERE person_age <= 100
"""
df_clean = pd.read_sql_query(query, conn)
df_clean
```

```python
df_clean.describe()
```

```python
# we cleaned credit_risk and now we have a new dataset (df_clean)
# dataset to sql
df_clean.to_sql('credit_risk_clean', conn, if_exists='replace', index=False)
```

```python
query = """
SELECT *
FROM credit_risk_clean
WHERE (person_age - person_emp_length) >= 16
"""
df_clean_2 = pd.read_sql_query(query, conn)
df_clean_2
```

```python
df_clean_2.describe()
```

```python
# we cleaned df_clean and now, we have new dataset (df_clean_2)
# dataset to sql
df_clean_2.to_sql('credit_risk_clean_2', conn, if_exists='replace', index=False)
```

```python
query = """
SELECT loan_grade,
AVG(loan_status) AS default_rate,
COUNT(*) AS loan_count
FROM credit_risk_clean_2
GROUP BY loan_grade
"""
df_loan_grade = pd.read_sql_query(query, conn)
df_loan_grade
```

```python
query = """
SELECT loan_intent,
AVG(loan_status) AS default_rate,
COUNT(*)AS loan_count
FROM credit_risk_clean_2
GROUP BY loan_intent
"""
df_loan_intent = pd.read_sql_query(query, conn)
df_loan_intent
```

```python
# PHASE 2 - SCIKIT LEARN PREDICTION
# Design Matrix formation-ish
X = df_clean_2.drop(columns = 'loan_status')
y = df_clean_2['loan_status']
# print(X)
type(y)
```

```python
# now we are going to make dummy variables for all the categorical variables
# we are going to drop the first one - keep it as baseline to avoid multicollinearity
X = pd.get_dummies(X, drop_first = True)
X
```
